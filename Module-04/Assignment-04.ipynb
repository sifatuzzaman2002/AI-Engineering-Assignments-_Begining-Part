# Q1. Simple Function with Parameters

```python
# Define a function to calculate the area of a rectangle
def calculate_area(length, width):
    return length * width

# Call the function
area = calculate_area(5, 4)

# Print the result
print("Area of the rectangle:", area)
```

# Q2. Greeting Function with Default Parameter

```python
# Define a greeting function with a default greeting
def greet(name, greeting="Hello"):
    print(greeting + ", " + name + "!")

# Call the function without passing a greeting
greet("Ayesha")

# Call the function with a custom greeting
greet("Ayesha", "Good Morning")
```

# Q3. Using Built-in math Module

```python
# Import the math module
import math

# Calculate square root of 25
square_root = math.sqrt(25)

# Calculate 2 raised to the power 3
power = math.pow(2, 3)

# Print the results
print("Square root of 25:", square_root)
print("2 raised to the power 3:", power)
```

# Q4. File Handling — Write and Read

```python
# Open the file in write mode
file = open("sample.txt", "w")

# Write content into the file
file.write("Welcome to Python Programming!")

# Close the file
file.close()

# Open the file in read mode
file = open("sample.txt", "r")

# Read the content
content = file.read()

# Print the content
print(content)

# Close the file
file.close()
```

# Q5. OOP Basics — Class and Object

```python
# Define the Student class
class Student:

    # Constructor method
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    # Method to display student information
    def display_info(self):
        print("Student Name:", self.name)
        print("Marks:", self.marks)


# Create an object of the Student class
student1 = Student("Ayesha", 92)

# Call the method
student1.display_info()
```

# List, Tuple, Set, Dictionary, enumerate() and zip()

```python
# Using enumerate()
fruits = ["Apple", "Banana", "Mango"]

print("Fruits:")
for index, fruit in enumerate(fruits):
    print(index, fruit)


# Using zip()
students = ["Ayesha", "Rahim", "Karim"]
marks = [92, 85, 88]

print("\nStudent Marks:")
for student, mark in zip(students, marks):
    print(student, mark)
```

```python
# List
fruits = ["Apple", "Banana", "Mango"]

# Tuple
colors = ("Red", "Green", "Blue")

# Set
numbers = {1, 2, 3, 4}

# Dictionary
student = {
    "name": "Ayesha",
    "marks": 92
}

print("List:", fruits)
print("Tuple:", colors)
print("Set:", numbers)
print("Dictionary:", student)
```
