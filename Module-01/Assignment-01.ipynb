# **Q1. Simple Calculator Program**

```python
# Write a Python program that performs basic arithmetic operations.
# 1.  Ask the user to input two numbers.
num1=input("Enter the Number")
print('The Entered First Number is :',num1)
num2=input("Enter the Number")
print('The Entered Second Number is :',num2)
```

```python
# 2. Convert both inputs into integers or floats.
num1=(float(num1))
num2=(float(num2))

print(f"{num1:.2f}")
print(f"{num2:.2f}")
 ### f = Formatted String Literal
```

```python
# Calculations
Addition=num1+num2  #Addition
Subtraction=num1-num2 #Subtraction
Multiplication=num1*num2 #Mul
Division=num1/num2 #Div
```

# **Complete Calculation with Print the output**

```python
print(f"The Addition of {num1} and {num2} is = {Addition:.2f}")
print(f"The Subtraction of {num1} and {num2} is = {Subtraction:.2f}")
print(f"The Multiplication of {num1} and {num2} is = {Multiplication:.2f}")
print(f"The Division of {num1} and {num2} is = {Division:.2f}")

#f use kora hoy karon f use korle string er vitore variable er value sorasori bosano jay
```

# **Q3. Temperature Converter**

```python
# Take the temperature in Celsius from the user.
celsius=input("Enter the temperature in Celsious: ")
# Convert to float.
celsius = float(celsius)
print(f"Entered temperature is : {celsius:.2f} in float data type")
```

```python
# Calculation
Farenhite = (celsius * 9/5) + 32
```

```python
# Print Fahrenheit with a message:
print(f"Temperature in Fahrenheit is: {Farenhite}°F ")
# Alt + 0176 → °
```
