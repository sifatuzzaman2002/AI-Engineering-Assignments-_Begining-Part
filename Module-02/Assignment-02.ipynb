# **Question 1: Loop Basics (for loop)**

```python
# Printing numbers from 1 to 10 using a for loop
for i in range(1, 11):
    print(i)
```

# **Question 2: Loop with Condition**

```python
# Print even numbers between 1 and 20
for i in range(1, 21):
    # Check Codition if the number is even
    if i % 2 == 0:
        print(i)
```

# **Question 3: String Manipulation**

```python
# Take a string input from the user
text = input("Enter a string: ")

# Print the string in uppercase
print("Uppercase:", text.upper())

# Print the length of the string
print("Length:", len(text))

# Print the string in reverse
print("Reversed:", text[::-1])
```

# **Question 4: String Processing with Loop**

```python
# Take a string input from the user
text = input("Enter a string: ")

# Initialize the vowel count
count = 0

# Check each character in the string
for char in text:
    # Check if the character is a vowel
    if char.lower() in "aeiou":
        count = count + 1

# Print the total number of vowels
print("Number of vowels:", count)
```
