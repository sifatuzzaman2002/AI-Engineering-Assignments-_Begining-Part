# **Question 1: Enumerate with List**

```python
# Question 1: Enumerate with List

fruits = ["apple", "banana", "cherry", "mango"]

for index, fruit in enumerate(fruits):
    print(index, fruit)
```

# **Question 2: Zip with Two Lists**

```python
# Question 2: Zip with Two Lists

students = ["Rahim", "Karim", "Ayesha"]
scores = [85, 90, 78]

for student, score in zip(students, scores):
    print(student, "scored", score)
```

# **Question 3: Tuple and Loop**

```python
# Question 3: Tuple and Loop

subjects = ("Math", "Physics", "Chemistry", "Biology", "English")

for subject in subjects:
    print(subject)
```

# **Question 4: Set Operations**

```python
# Question 4: Set Operations

set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7}

# Print common elements
print("Common elements:", set1.intersection(set2))

# Print unique elements of set1
print("Unique elements of set1:", set1.difference(set2))
```

# **Question 5: Dictionary with Enumerate**

```python
# Question 5: Dictionary with Enumerate

marks = {
    "Math": 80,
    "Physics": 75,
    "Chemistry": 85
}

for number, (subject, mark) in enumerate(marks.items(), start=1):
    print(number, subject + ":", mark)
```
